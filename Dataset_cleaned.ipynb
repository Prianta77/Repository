```python
import pandas as pd
import re
from google.colab import files

# 1. Load Dataset
df = pd.read_csv('Twitter_Emotion_Dataset.csv')
print(f"Jumlah data awal: {len(df)}")

# 2. Hapus Data Duplikat
df = df.drop_duplicates(subset=['tweet'])
print(f"Jumlah data setelah hapus duplikat: {len(df)}")

# 3. Fungsi untuk Cleaning Teks
def bersihkan_teks(teks):
    teks = str(teks)

    # Menghapus tag [USERNAME] dan [URL]
    teks = teks.replace('[USERNAME]', '')
    teks = teks.replace('[URL]', '')

    # Mengubah menjadi huruf kecil (lowercase)
    teks = teks.lower()

    # Menghapus angka dan tanda baca (hanya menyisakan huruf a-z dan spasi)
    teks = re.sub(r'[^a-z\s]', '', teks)

    # Menghapus spasi yang berlebihan (misal dari sisa penghapusan tag)
    teks = re.sub(r'\s+', ' ', teks).strip()

    return teks

# 4. Terapkan fungsi cleaning ke kolom 'tweet'
df['tweet'] = df['tweet'].apply(bersihkan_teks)

# 5. Hapus baris yang teksnya menjadi kosong setelah dibersihkan
df = df[df['tweet'] != '']
print(f"Jumlah data akhir setelah cleaning penuh: {len(df)}")

# Cek 5 data teratas setelah dibersihkan
print("\n=== Sample Data Bersih ===")
print(df.head())

# 6. Simpan hasil cleaning ke file CSV baru
nama_file_baru = 'Twitter_Emotion_Cleaned.csv'
df.to_csv(nama_file_baru, index=False)
print(f"\nData berhasil disimpan sebagai '{nama_file_baru}'")

# 7. Otomatis download file ke laptop
files.download(nama_file_baru)
```
